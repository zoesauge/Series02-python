## Equation of a sine wave
$$y(t) = A \sin(\omega t + \phi)$$

Where:
* $y(t)$: instantaneous value (its depends of what is measured)
* $A$: peak amplitude (its depends of what is measured)
* $\omega$: angular frequency ($\text{rad}\cdot\text{s}^{-1}$)
* $t$: time ($\text{s}$)
* $\phi$: initial phase or phase shift ($\text{rad}$)

For the first plot, we use a frequency $f = 1\text{ Hz}$, which gives the equation:
$$s(t) = \sin(2\pi t)$$

```python
# A. import the needed libraries
import numpy as np
import matplotlib.pyplot as plt
# B. Create a numpy array called t (for time) that contains 1000 values from a 0 to 1
t = np.linspace(0, 1, 1000)
# C. Create a numpy array called s that contains the values of a sine function  of t
s = np.sin(2 * np.pi * t)
# D. Plot s as a function of t
plt.plot(t, s)
# E. Add a title to the plot
plt.title('Sine Function')
# F. Add labels to the x axis
plt.xlabel('Time (s)')
# G. Add labels to the y axis
plt.ylabel('Sine')
# H. Add a grid to the plot 
plt.grid(True)
# I. Save the plot as a png file
plt.savefig('sine_function.png')
# Display the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sinusoid(t, amplitude=1.0, frequency=1.0, phase=0.0):
    """
    Computes and returns a NumPy array representing a sinusoidal signal.
    
    Parameters:
    - t (array): NumPy array representing time.
    - amplitude (float): Maximum amplitude of the sinusoid.
    - frequency (float): Frequency of the signal in Hertz (Hz).
    - phase (float): Initial phase in radians.
    
    Returns:
    - x (array): NumPy array containing the computed signal values.
    """
    x = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    return x

# Time axis definition (1000 points between 0 and 1 second)
#Nous définissons un vecteur temporel haute résolution pour un meilleur rendu des signaux (sans pixellisation)
t = np.linspace(0, 1, 1000)

# Creating two numpy arrays s1 and s2 with two different frequencies
#Nous générons deux tableaux distincts avec des fréquences différentes pour comparer l'impact de la fréquence sur la vitesse d'oscillation de l'onde.
s1 = sinusoid(t, frequency=1.0)  # Frequency for Sine 1
s2 = sinusoid(t, frequency=2.0)  # Frequency for Sine 2

# Creating the figure for plotting
plt.figure(figsize=(8, 4))

# Plotting s1 and s2 as a function of t
# Nous traceons les deux signaux avec des styles distincts (pour les différencier)
plt.plot(t, s1, label='Sinus 1', color='tab:blue')
plt.plot(t, s2, label='Sinus 2', color='tab:orange')

# Adding readability elements (title, axes, grid, and legend)
# Pour rendre le graphique plus lisible
plt.title('Sine and Cosine Functions')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid(True)
plt.legend()

# Saving the plot as a PDF file
plt.savefig('sine_cosine_function.pdf', format='pdf', bbox_inches='tight')
plt.show()
```

## Even numbers and odd numbers 
An even number is an integer that is a multiple of 2. This means that when divided by 2, it leaves no remainder

```python
import numpy as np

# creation of a series number from 0 to 10
series = np.arange(0, 11)

# We loop through each nomber to test if it's enven or odd
for number in series:
    if number % 2 == 0:
        print(f"{number} is even")
    else:
        print(f"{number} is odd")
        
```

A multiple of a number $m$ is an integer obtained by multiplying $m$ by another integer. In other words, the remainder when it is divided by $m$ is equal to 0 ($n \pmod m = 0$).

```python
import numpy as np

# We create a sequence of numbers from 0 to 20 to test them
numbers = np.arange(0, 21)

# We loop through each number to check if it is a multiple of 3
for number in numbers:
    # We check if the remainder of the division by 3 is equal to 0
    if number % 3 == 0:
        print(f"{number} is a multiple of 3")
    else:
        print(f"{number} is not a multiple of 3")

# We test each number from 0 to 20 to check if it is a multiple of 5
for number in numbers:
    if number % 5 == 0:
        print(f"{number} is a multiple of 5")
    else:
        print(f"{number} is not a multiple of 5")
```

```python
# We define a function to check if a specific number is a multiple of m. Function that takes a number and m, returning True or False
def is_multiple_of(number, m):
    return number % m == 0
# Example of use with m = 3
print(is_multiple_of(9, 3))  # True
print(is_multiple_of(10, 3))  # False
```

### Application à une série complète de nombres

L'objectif ici est de vérifier d'un seul coup, pour toute une série de nombres, lesquels sont des multiples de $m$. La fonction renvoie une liste de valeurs logiques : `True` si c'est un multiple, et `False` sinon.

```python
# We define a function that takes a series of numbers and m,  and returns a list of booleans indicating if each number is a multiple of m
def get_multiples_boolean_list(series_numbers, m):
    # We use a list comprehension to check the condition for every number in the series
    return [is_multiple_of(num, m) for num in series_numbers]

# Example with numbers from 0 to 10 and m = 3
numbers_series = np.arange(0, 11)
print(get_multiples_boolean_list(numbers_series, 3))
```

### Prime or not ?
Un nombre premier est un entier naturel supérieur à 1 qui n'admet que deux diviseurs distincts : 1 et lui-même.

```python
# We create a sequence of integers from 0 to 20
numbers_prime = np.arange(0, 21)

# We define a function to check if a number is prime using our previous multiple-checking logic
def is_prime(n):
    # A number is prime if it is not a multiple of any integer between 2 and itself-minus-1
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

# We find all prime numbers in our series
primes = [n for n in numbers_prime if is_prime(n)]
print("Nombres premiers de 0 à 20 :", primes)
```
