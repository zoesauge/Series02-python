```python
import numpy as np
import matplotlib.pyplot as plt

def sinusoid(t, amplitude=1.0, frequency=1.0, phase=0.0):
    """
    Computes and returns a NumPy array representing a sinusoidal signal.
    
    Parameters:
    - t (array): NumPy array representing time.
    - amplitude (float): Maximum amplitude of the sinusoid.
    - frequency (float): Frequency of the signal in Hertz (Hz).
    - phase (float): Initial phase in radians.
    
    Returns:
    - x (array): NumPy array containing the computed signal values.
    """
    x = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    return x

# Time axis definition (1000 points between 0 and 1 second)
#Nous définissons un vecteur temporel haute résolution pour un meilleur rendu des signaux (sans pixellisation)
t = np.linspace(0, 1, 1000)

# Creating two numpy arrays s1 and s2 with two different frequencies
#Nous générons deux tableaux distincts avec des fréquences différentes pour comparer l'impact de la fréquence sur la vitesse d'oscillation de l'onde.
s1 = sinusoid(t, frequency=1.0)  # Frequency for Sine 1
s2 = sinusoid(t, frequency=2.0)  # Frequency for Sine 2

# Creating the figure for plotting
plt.figure(figsize=(8, 4))

# Plotting s1 and s2 as a function of t
# Nous traceons les deux signaux avec des styles distincts (pour les différencier)
plt.plot(t, s1, label='Sinus 1', color='tab:blue')
plt.plot(t, s2, label='Sinus 2', color='tab:orange')

# Adding readability elements (title, axes, grid, and legend)
# Pour rendre le graphique plus lisible
plt.title('Sine and Cosine Functions')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid(True)
plt.legend()

# Saving the plot as a PDF file
plt.savefig('sine_cosine_function.pdf', format='pdf', bbox_inches='tight')
plt.show()
```
