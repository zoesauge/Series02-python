# Serie 02 : to code and plot sinusoidql signals 
The goal of this notebook is to creat a sinusoidal signals 

## Equation of a sine wave
$$y(t) = A \sin(\omega t + \phi)$$


```python
# A. import the needed libraries
import numpy as np #numpy is a library for numerical computations in Python
import matplotlib.pyplot as plt #matplotlib is a library for creating visualizations in Python
# B. Create a numpy array called t (for time) that contains 1000 values from a 0 to 1
t = np.linspace(0, 1, 1000)
# C. Create a numpy array called s that contains the values of a sine function  of t
s = np.sin(2 * np.pi * t)
# D. Plot s as a function of t
plt.plot(t, s)
# E. Add a title to the plot
plt.title('Sine Function')
# F. Add labels to the x axis
plt.xlabel('Time')
# G. Add labels to the y axis
plt.ylabel('Sine')
# H. Add a grid to the plot 
plt.grid(True)
# I. Save the plot as a png file
plt.savefig('sine_function.png')
# Display the plot
plt.show()
```
