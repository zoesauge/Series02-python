# Serie 02 : to code and plot sinusoidql signals 
The goal of this notebook is to creat a sinusoidal signals 

## Equation of a sine wave
$$y(t) = A \sin(\omega t + \phi)$$


```python
# A. import the needed libraries
import numpy as np #numpy is a library for numerical computations in Python
import matplotlib.pyplot as plt #matplotlib is a library for creating visualizations in Python
# B. Create a numpy array called t (for time) that contains 1000 values from a 0 to 1
t = np.linspace(0, 1, 1000)
# C. Create a numpy array called s that contains the values of a sine function  of t
s = np.sin(2 * np.pi * t)
# D. Plot s as a function of t
plt.plot(t, s)
# E. Add a title to the plot
plt.title('Sine Function')
# F. Add labels to the x axis
plt.xlabel('Time')
# G. Add labels to the y axis
plt.ylabel('Sine')
# H. Add a grid to the plot 
plt.grid(True)
# I. Save the plot as a png file
plt.savefig('sine_function.png')
# Display the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Définition du temps et du premier signal (le sinus de l'étape précédente)
t = np.linspace(0, 1, 1000)
s = np.sin(2 * np.pi * t)
plt.figure(figsize=(8, 4)) #figsize is used to set the size of the figure, here 8 inches wide and 4 inches tall
plt.plot(t, s, label='Sine Function')

# A. Add a second plot of cosine function of t on the same figure
c = np.cos(2 * np.pi * t)
# B. Add a legend to the plot
plt.plot(t, c, label='Cosine Function')
plt.legend(loc='lower left')
# C. Add a second y axis on the plot
ax2 = plt.gca().twinx()
# D. Add a second label to the y axis
ax2.set_ylabel('Cosine')
# E. Add a title to the plot
plt.title('Sinusoidal Function')
# F. Add labels to the x axis
plt.xlabel('Time')
# G. Add labels to the y axis
plt.ylabel('Amplitude')
# H. Add a grid to the plot 
plt.grid(True)
# I. Save the plot as a pdf file
plt.savefig('sine_cosine_function.pdf') 
plt.show()
```
